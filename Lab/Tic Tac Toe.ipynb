```python
import math

class TicTacToe:
    def __init__(self):
        # 3x3 board initialized with empty spaces
        self.board = [[' ' for _ in range(3)] for _ in range(3)]
        self.human = 'X'
        self.ai = 'O'

    def print_board(self):
        """Displays the board in a clean terminal grid."""
        print("\n")
        for row_idx, row in enumerate(self.board):
            print(" " + " | ".join(row) + " ")
            if row_idx < 2:
                print("---|---|---")
        print("\n")

    def print_positions(self):
        """Displays position numbers (1-9) for player guidance."""
        print("\nBoard Position Guide:")
        print(" 1 | 2 | 3 ")
        print("---|---|---")
        print(" 4 | 5 | 6 ")
        print("---|---|---")
        print(" 7 | 8 | 9 \n")

    def is_moves_left(self):
        """Checks if there are open spots on the board."""
        for row in self.board:
            if ' ' in row:
                return True
        return False

    def evaluate(self):
        """
        Evaluates current board state.
        Returns +10 if AI wins, -10 if Human wins, 0 for ongoing or draw.
        """
        # Check Rows & Columns
        for i in range(3):
            if self.board[i][0] == self.board[i][1] == self.board[i][2] != ' ':
                return 10 if self.board[i][0] == self.ai else -10
            if self.board[0][i] == self.board[1][i] == self.board[2][i] != ' ':
                return 10 if self.board[0][i] == self.ai else -10

        # Check Diagonals
        if self.board[0][0] == self.board[1][1] == self.board[2][2] != ' ':
            return 10 if self.board[0][0] == self.ai else -10
        if self.board[0][2] == self.board[1][1] == self.board[2][0] != ' ':
            return 10 if self.board[0][2] == self.ai else -10

        return 0

    def minimax(self, depth, is_maximizing):
        """
        Minimax recursive function to calculate optimal move scores.
        """
        score = self.evaluate()

        # Terminal States
        if score == 10:
            return score - depth  # Prefer faster wins
        if score == -10:
            return score + depth  # Prefer slower losses
        if not self.is_moves_left():
            return 0

        if is_maximizing:
            best = -math.inf
            for r in range(3):
                for c in range(3):
                    if self.board[r][c] == ' ':
                        self.board[r][c] = self.ai
                        best = max(best, self.minimax(depth + 1, False))
                        self.board[r][c] = ' '
            return best
        else:
            best = math.inf
            for r in range(3):
                for c in range(3):
                    if self.board[r][c] == ' ':
                        self.board[r][c] = self.human
                        best = min(best, self.minimax(depth + 1, True))
                        self.board[r][c] = ' '
            return best

    def find_best_move(self):
        """Finds the optimal move for AI using Minimax."""
        best_val = -math.inf
        best_move = (-1, -1)

        for r in range(3):
            for c in range(3):
                if self.board[r][c] == ' ':
                    self.board[r][c] = self.ai
                    move_val = self.minimax(0, False)
                    self.board[r][c] = ' '

                    if move_val > best_val:
                        best_move = (r, c)
                        best_val = move_val

        return best_move

    def play(self):
        """Main game loop."""
        print("Welcome to Tic Tac Toe! You are 'X' and the AI is 'O'.")
        self.print_positions()

        while self.is_moves_left():
            # Player Turn
            try:
                move = int(input("Enter your move (1-9): ")) - 1
                r, c = divmod(move, 3)
                if not (0 <= move <= 8) or self.board[r][c] != ' ':
                    print("Invalid move! Pick an empty spot from 1 to 9.")
                    continue
            except ValueError:
                print("Invalid input! Enter a number between 1 and 9.")
                continue

            self.board[r][c] = self.human
            self.print_board()

            if self.evaluate() == -10:
                print("Congratulations! You won!")
                return
            if not self.is_moves_left():
                break

            # AI Turn
            print("AI is thinking...")
            ai_r, ai_c = self.find_best_move()
            self.board[ai_r][ai_c] = self.ai
            self.print_board()

            if self.evaluate() == 10:
                print("AI Wins! Better luck next time.")
                return

        print("It's a Draw!")

# ================= Run the Game =================
game = TicTacToe()
game.play()
2
```
