```python
import math

def solve_tsp(cost_matrix):
    """
        Solves TSP using the Held-Karp dynamic programming algorithm.

                :param cost_matrix: 2D list where cost_matrix[i][j] is the cost/distance from city i to city j
                    :return: (min_cost, path) - minimum path cost and ordered list of visited cities starting/ending at city 0
                        """
    n = len(cost_matrix)

    # Memoization table: memo[(mask, current_city)] = (min_cost, prev_city)
    # mask represents the set of visited cities as a bitmask (e.g., bit 0 set -> city 0 visited)
    memo = {}

    def tsp(mask, u):
        # Base case: All cities visited, return to start city (0)
        if mask == (1 << n) - 1:
            return cost_matrix[u][0], 0

        state = (mask, u)
        if state in memo:
            return memo[state]

        min_cost = float('inf')
        best_next_city = None

        # Try visiting every unvisited city 'v'
        for v in range(n):
            if not (mask & (1 << v)):  # City v is not visited
                cost, _ = tsp(mask | (1 << v), v)
                total_cost = cost_matrix[u][v] + cost
                if total_cost < min_cost:
                    min_cost = total_cost
                    best_next_city = v

        memo[state] = (min_cost, best_next_city)
        return memo[state]

    # Start tour from city 0 with mask = 1 (only city 0 visited)
    min_cost_final, _ = tsp(1, 0)

    # Reconstruct the optimal path
    path = [0]
    curr_city = 0
    curr_mask = 1

    while curr_mask != (1 << n) - 1:
        _, next_city = memo[(curr_mask, curr_city)]
        path.append(next_city)
        curr_mask |= (1 << next_city)
        curr_city = next_city

    path.append(0)  # Return to start

    return min_cost_final, path


# ================= Example Usage =================
# Distance/cost matrix for 4 cities
#     City 0  City 1  City 2  City 3
distance_matrix = [
        [ 0,     10,     15,     20],  # City 0
        [10,      0,     35,     25],  # City 1
        [15,     35,      0,     30],  # City 2
        [20,     25,     30,      0]   # City 3
    ]

cost, path = solve_tsp(distance_matrix)

print(f"Minimum Tour Cost: {cost}")
print(f"Optimal Path: {' -> '.join(f'City {c}' for c in path)}")
```
