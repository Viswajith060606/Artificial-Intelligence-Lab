```python
import math

def alphabeta(node_index, depth, alpha, beta, is_maximizing, scores, branching_factor):
    """
    Minimax search with Alpha-Beta Pruning.

    :param node_index: Index of current node in the scores array
    :param depth: Remaining search depth
    :param alpha: Maximizer's best score guarantee
    :param beta: Minimizer's best score guarantee
    :param is_maximizing: True if turn is Maximizer, False for Minimizer
    :param scores: Array of leaf node values
    :param branching_factor: Number of children per node
    """
    # Base case: Leaf node reached
    if depth == 0:
        return scores[node_index]

    if is_maximizing:
        max_eval = -math.inf
        for i in range(branching_factor):
            child_index = node_index * branching_factor + i
            eval_score = alphabeta(child_index, depth - 1, alpha, beta, False, scores, branching_factor)
            max_eval = max(max_eval, eval_score)
            alpha = max(alpha, eval_score)

            # Beta Cutoff (Prune remaining children)
            if beta <= alpha:
                break
        return max_eval

    else:
        min_eval = math.inf
        for i in range(branching_factor):
            child_index = node_index * branching_factor + i
            eval_score = alphabeta(child_index, depth - 1, alpha, beta, True, scores, branching_factor)
            min_eval = min(min_eval, eval_score)
            beta = min(beta, eval_score)

            # Alpha Cutoff (Prune remaining children)
            if beta <= alpha:
                break
        return min_eval


# Example Usage:
# Leaf nodes values corresponding to a binary tree of depth 3 (2^3 = 8 leaves)
leaf_scores = [3, 5, 6, 9, 1, 2, 0, -1]

# Start at root (index 0), depth 3, alpha = -inf, beta = +inf
optimal_score = alphabeta(
    node_index=0,
    depth=3,
    alpha=-math.inf,
    beta=math.inf,
    is_maximizing=True,
    scores=leaf_scores,
    branching_factor=2
)

print(f"Optimal Score: {optimal_score}")
```
